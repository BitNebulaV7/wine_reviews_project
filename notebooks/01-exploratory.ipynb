```python
import pandas as pd
from pathlib import Path
import matplotlib.pyplot as plt

RAW = Path("../data/raw/winemag-data-130k-v2.csv")

def load_data(filepath, dtypes=None):
    try:
        df = pd.read_csv(filepath, dtype=dtypes)
        print("✅ File loaded successfully.")
        print("🔍 Missing values per column:")
        print(df.isnull().sum())
        return df
    except Exception as e:
        print(f"❌ Error loading file: {e}")
        return None
```

```python
df = load_data(RAW)
df.head()
```

```python
df.shape
df.info()
```

```python
missing_values = df.isnull().sum()
missing_percent =  (missing_values / len(df)) * 100

missing_df = pd.DataFrame({
    'missing count': missing_values,
    'missing percent': missing_percent.round(2)
}).sort_values(by='missing percent', ascending=False)

print(missing_df)
```

```python
df.drop(columns=['region_2','taster_twitter_handle'],inplace=True)
```

```python
median_price = df['price'].median()
df['price'] = df['price'].fillna(median_price)

mode_country = df['country'].mode()[0]
df['country'] = df['country'].fillna(mode_country)

mode_province = df['province'].mode()[0]
df['province'] = df['province'].fillna(mode_province)

mode_variety = df['variety'].mode()[0]
df['variety'] = df['variety'].fillna(mode_variety)

df['taster_name'] = df['taster_name'].fillna('Unknown')
```
