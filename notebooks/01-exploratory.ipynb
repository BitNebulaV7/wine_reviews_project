```python
import pandas as pd
from pathlib import Path
import matplotlib.pyplot as plt

RAW = Path("../data/raw/winemag-data-130k-v2.csv")

def load_data(filepath, dtypes=None):
    try:
        df = pd.read_csv(filepath, dtype=dtypes)
        print("✅ File loaded successfully.")
        print("🔍 Missing values per column:")
        print(df.isnull().sum())
        return df
    except Exception as e:
        print(f"❌ Error loading file: {e}")
        return None
```

```python
df = load_data(RAW)
df.head()
```

```python
def summarise_df(df):
    num_cols,num_rows = df.shape
    print(f"🔢 Rows: {num_rows}, Columns: {num_cols}")

    missing_count = df.isnull().sum()
    missing_percent = (missing_count / len(df)) * 100

    unique_counts = df.nunique()

    summary = pd.DataFrame({
        'missing count' : missing_count,
        'missing %' : missing_percent.round(2),
        'unique Values' : unique_counts

    }).sort_values(by='missing %', ascending=False)

    return summary
```

```python
summary_table = summarise_df(df)
summary_table.head(10)
```

```python
df.drop(columns=['region_2','taster_twitter_handle'],inplace=True)
```

```python
median_price = df['price'].median()
df['price'] = df['price'].fillna(median_price)

mode_country = df['country'].mode()[0]
df['country'] = df['country'].fillna(mode_country)

mode_province = df['province'].mode()[0]
df['province'] = df['province'].fillna(mode_province)

mode_variety = df['variety'].mode()[0]
df['variety'] = df['variety'].fillna(mode_variety)

df['taster_name'] = df['taster_name'].fillna('Unknown')
```
